## Reusable RAG Pipeline Notes

### Why this notebook matters
In earlier lessons, we built Retrieval-Augmented Generation (RAG) one piece at a time:
1. Search over FAQ documents
2. Prompt construction with retrieved context
3. LLM response generation

That approach is great for learning, but repetitive for daily use. In this notebook, we switch to a reusable workflow by importing helper modules.

### Reusable modules used here
- `ingest.py`: loads FAQ data and builds the search index
- `rag_helper.py`: encapsulates the core RAG flow (retrieve -> build prompt -> call LLM)

### Study takeaway
This structure keeps notebooks focused on experiments and analysis instead of boilerplate setup code.

```python
# Load environment variables from .env (e.g., OPENAI_API_KEY).
from dotenv import load_dotenv
load_dotenv()
```

```python
# Import reusable helpers and the OpenAI client.
from ingest import load_faq_data, build_index
from rag_helper import RAGBase
from openai import OpenAI
```

### Step 1: Build the Knowledge Base
We load FAQ documents and create a searchable index. This index is what the assistant uses to retrieve relevant context before generating an answer.

```python
# Load FAQ records and build a search index for retrieval.
documents = load_faq_data()
index = build_index(documents)
```

### Step 2: Initialize the RAG Assistant
Here we create an OpenAI client and pass it to `RAGBase` together with the retrieval index. From this point, calling `assistant.rag(...)` runs the complete pipeline.

```python
# Create the LLM client and wire it into the reusable RAG assistant.
openai_client = OpenAI()

assistant = RAGBase(
    index=index,
    llm_client=openai_client,
)
```

### Step 3: Validate the Assistant with Sample Questions
We now test the reusable pipeline with representative learner questions. The goal is to confirm that retrieval and response generation work reliably across common FAQ intents.

```python
# Test 1: Enrollment timing question.
answer1 = assistant.rag("I just discovered the course. Can I join now?")
print(answer1)
```

```python
# Test 2: Beginner-friendliness question.
answer2 = assistant.rag("Is the course for beginners?")
print(answer2)
```

```python
# Test 3: Prerequisites question.
answer3 = assistant.rag("What do I need to know to join the course?")
print(answer3)
```

```python
answer4 = assistant.rag("Can I still join the course after it started?")
print(answer4)
```
